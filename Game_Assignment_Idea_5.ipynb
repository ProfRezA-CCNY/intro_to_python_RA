```python
import random

health = 100
infection = 50

def antibodies():
    damage = int(random.gauss(10,15))
    if damage<0: damage = 0
    return damage

def t_cells():
    damage = random.randint(5,30)
    return damage

def rest():
    heal = random.randint(5,20)
    return heal

def display(hp =100, inf_hp = 50, split = 10):
    str_hp = "Health:"
    if split<len(str_hp):
        text_length = len(str_hp)
    else:
        text_length = split

    str_infection = "Infection:"
    print(str_hp,end = " ")
    print(" "*(text_length-len(str_hp))," | ", end=" ")
    print(str_infection)
    num_rows = max(hp//split, inf_hp//split)+1
    marker = "*"
    hp_row = 0
    hp_row_max = hp//split
    hp_remainder = hp%split
    inf_hp_row_max = inf_hp//split
    inf_hp_row = 0
    inf_hp_remainder =inf_hp%split
    for _ in range(num_rows):
        if hp_row<hp_row_max:
            print(marker*split," "*(text_length-split)," | ", end = " ")
        elif hp_row == hp_row_max:
            print(marker*(hp_remainder)," "*(text_length-hp_remainder)," | ", end = " ")
        else:
            print(" "*text_length," | ", end = " ")
        if inf_hp_row<inf_hp_row_max:
            print(marker*split, end ="\n")
        elif inf_hp_row == inf_hp_row_max:
            print(marker*inf_hp_remainder, end ="\n")
        else:
            print('')
        hp_row += 1
        inf_hp_row += 1

while health > 0 and infection > 0:

    print("\nHealth:", health)
    print("Infection:", infection)

    print("Choose an action:")
    print("1 - Release Antibodies")
    print("2 - Activate T Cells")
    print("3 - Rest")

    choice = input("> ")

    if choice == "1":
        infection -= antibodies()
    elif choice == "2":
        infection -= t_cells()
    elif choice == "3":
        health += rest()

    attack = random.randint(5,25)
    health -= attack
    print("The infection attacks!\n")
    display(health, infection)

if infection <= 0:
    print("You defeated the infection!")
else:
    print("The infection won...")
```
