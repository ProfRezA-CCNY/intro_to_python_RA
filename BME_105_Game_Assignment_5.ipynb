#**Assignment: Build a “Fight the Infection” Python Game**
## <u>Objective</u>

Create a simple text-based Python game where the player acts as the immune system fighting an infection in the body. The player will make decisions to eliminate pathogens while keeping the body healthy.

This assignment introduces:

Variables

Conditionals (if/else)

Loops

Functions

Random numbers

Scenario

* Variables
* Conditionals (if/else)
* Loops
* Functions
* Random Numbers
* Scenario
<hr>

**A harmful bacterium has entered the bloodstream. You are the immune system, and your goal is to eliminate the infection before the patient's health reaches zero.**

Each round:

1. The infection attacks the body.

2. The player chooses an immune response.

**Game Rules:**</p>
<u>Starting Conditions</u>
<p/>Patient Health = 100
<p/>Infection Strength = 50
<u><p>Player Actions:</p></u>
<p/>Each turn the player chooses:

1. **Release Antibodies:**
    * Reduces infection by **10–20** points
2. **Activate T-Cells:**
    * Reduces infection by **5–30** points
3. **Rest the Body:**
    * Increases health by **5–15** points

<u><p/>Infection Attack</u>
<p>After each turn, the infection attacks:</p>

* Patient health decreases **5–15** points

## <u>Win/Loss Conditions</u>

**Player Wins**: Infection strength reaches 0

**Player Loses**: Patient health reaches 0
